## Key takeaways

- Security, privacy, and compliance are three different questions, and a service can satisfy one while failing another.
- A prompt injection pattern list is a first layer, not a complete defense, and retrieved documents deserve the same scrutiny as user input.
- Redaction has to happen before data is stored or logged, never after, since a written log line cannot be un-logged.
- Secrets belong in environment variables only, masked whenever printed, never hardcoded or logged in full.
- Least privilege means a leaked key should only ever expose what its role actually needs.
- A daily budget and a scrape detector catch abuse patterns a per-minute rate limit alone cannot.
- Data should be deleted on a real schedule, and a user's deletion request should actually be honorable on demand.
- This is teaching code. A real system needs a professional security review and likely legal review before launch.
